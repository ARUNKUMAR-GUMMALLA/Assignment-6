1. Create a simple page with some div tags and show different ways to add CSS as well as what happens
when you target the same elements with inline, internal, and external CSS. Also, utilize comments in the
project where required.

from IPython.display import HTML, display

html = """
< !DOCTYPE html>
< html>
< head>
    < title>CSS Example< /title>

    < style>
        /* Internal CSS */

        .internal {
            background-color: lightblue;
            color: blue;
            padding: 15px;
            margin: 10px;
        }

        .same {
            background-color: yellow;
            color: black;
        }
    < /style>
< /head>

< body>

    < h1>Different Ways to Add CSS< /h1>

    < div class="external">
        This div uses external CSS.
    < /div>

    < div class="internal">
        This div uses internal CSS.
    < /div>

    < div class="external"
         style="background-color: purple; color: white;">
        This div uses inline CSS.
    < /div>

    < div class="same"
         style="background-color: red; color: white;">
        This div shows CSS priority.
    < /div>

< /body>
< /html>
"""

html = html.replace("< !", "<!").replace("< /", "</").replace("< ", "<")

display(HTML(html))

```python
from IPython.display import HTML, display

html = """
< !DOCTYPE html>
< html>
< head>
    < title>CSS Example< /title>

    < style>
        /* Internal CSS */

        .internal {
            background-color: lightblue;
            color: blue;
            padding: 15px;
            margin: 10px;
        }

        .same {
            background-color: yellow;
            color: black;
        }
    < /style>
< /head>

< body>

    < h1>Different Ways to Add CSS< /h1>

    < div class="external">
        This div uses external CSS.
    < /div>

    < div class="internal">
        This div uses internal CSS.
    < /div>

    < div class="external"
         style="background-color: purple; color: white;">
        This div uses inline CSS.
    < /div>

    < div class="same"
         style="background-color: red; color: white;">
        This div shows CSS priority.
    < /div>

< /body>
< /html>
"""

html = html.replace("< !", "<!").replace("< /", "</").replace("< ", "<")

display(HTML(html))
```
