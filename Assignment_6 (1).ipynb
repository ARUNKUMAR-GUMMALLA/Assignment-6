# Starting with CSS


1. Create a simple page with some div tags and show different ways to add CSS as well as what happens
when you target the same elements with inline, internal, and external CSS. Also, utilize comments in the
project where required.

```python
from IPython.display import HTML, display

html = """
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">

    <title>CSS Example</title>

    <!-- Internal CSS -->
    <style>
        /* Styling using internal CSS */
        .internal {
            background-color: lightblue;
            color: blue;
            padding: 15px;
            margin: 10px;
        }

        /* Same class is also styled in external CSS */
        .same {
            background-color: yellow;
            color: black;
        }
    </style>

    <!-- External CSS -->
    <style>
        /* External CSS */
        .external {
            background-color: lightgreen;
            color: darkgreen;
            padding: 15px;
            margin: 10px;
            border: 2px solid green;
        }

        .same {
            background-color: orange;
            color: black;
            padding: 15px;
            margin: 10px;
            border: 2px solid orange;
        }
    </style>
</head>

<body>

    <h1>Different Ways to Add CSS</h1>

    <!-- External CSS -->
    <div class="external">
        This div uses external CSS.
    </div>

    <!-- Internal CSS -->
    <div class="internal">
        This div uses internal CSS.
    </div>

    <!-- Inline CSS -->
    <div class="external" style="background-color: purple; color: white;">
        This div uses inline CSS.
    </div>

    <!-- Same element targeted by different CSS methods -->
    <div class="same" style="background-color: red; color: white;">
        This div shows CSS priority.
    </div>

</body>
</html>
"""
```

```python
display(HTML(html))
```

2.  Build an HTML page with multiple paragraphs, each assigned a unique class name. Write CSS rules using
class selectors to apply distinct styling to each paragraph. Follow the BEM naming convention and
explain how you've named the classes

```python
from IPython.display import HTML, display

html = """
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <title>BEM Class Selectors</title>

    <style>
        /* Block: paragraph section */
        .paragraph-section {
            width: 80%;
            margin: 30px auto;
            font-family: Arial, sans-serif;
        }

        /* Element: introduction paragraph */
        .paragraph-section__intro {
            background-color: lightblue;
            color: darkblue;
            padding: 15px;
            border-left: 5px solid blue;
        }

        /* Element: information paragraph */
        .paragraph-section__info {
            background-color: lightgreen;
            color: darkgreen;
            padding: 15px;
            border-left: 5px solid green;
        }

        /* Element: warning paragraph */
        .paragraph-section__warning {
            background-color: lightyellow;
            color: darkorange;
            padding: 15px;
            border-left: 5px solid orange;
        }

        /* Element: conclusion paragraph */
        .paragraph-section__conclusion {
            background-color: lightgray;
            color: black;
            padding: 15px;
            border-left: 5px solid gray;
        }
    </style>
</head>

<body>

    <div class="paragraph-section">

        <p class="paragraph-section__intro">
            This is an introduction paragraph. It uses a unique class
            and has its own styling.
        </p>

        <p class="paragraph-section__info">
            This paragraph contains some useful information.
            It has a different background and text color.
        </p>

        <p class="paragraph-section__warning">
            This is a warning paragraph. The orange color makes
            the warning easy to notice.
        </p>

        <p class="paragraph-section__conclusion">
            This is the conclusion paragraph. It has a simple
            style to finish the page.
        </p>

    </div>

</body>
</html>
"""

display(HTML(html))
```

3. Develop an HTML form with various input elements. Use CSS to style the form, including setting
background colors for input fields. Create a custom color palette for the form elements, and demonstrate
how to apply opacity to one of the form sections.

```python
from IPython.display import HTML, display

html = """
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">

    <title>Styled HTML Form</title>

    <style>
        /* Custom color palette */
        :root {
            --dark: #243447;
            --blue: #3498db;
            --light-blue: #eaf4fb;
            --white: #ffffff;
            --green: #2ecc71;
            --border: #b8c7d1;
        }

        body {
            font-family: Arial, sans-serif;
            background-color: var(--light-blue);
            margin: 0;
            padding: 30px;
        }

        .form {
            width: 400px;
            margin: auto;
            background-color: var(--white);
            padding: 25px;
            border-radius: 8px;
        }

        .form h2 {
            text-align: center;
            color: var(--dark);
        }

        .form label {
            display: block;
            margin-top: 15px;
            color: var(--dark);
        }

        .form input,
        .form select,
        .form textarea {
            width: 100%;
            padding: 10px;
            margin-top: 5px;
            box-sizing: border-box;
            border: 1px solid var(--border);
            border-radius: 5px;
            background-color: var(--light-blue);
        }

        .form input[type="radio"],
        .form input[type="checkbox"] {
            width: auto;
        }

        /* Section with opacity */
        .form-section {
            background-color: var(--blue);
            color: var(--white);
            padding: 15px;
            margin-top: 20px;
            border-radius: 5px;
            opacity: 0.7;
        }

        .form button {
            width: 100%;
            margin-top: 20px;
            padding: 12px;
            border: none;
            border-radius: 5px;
            background-color: var(--green);
            color: var(--white);
            cursor: pointer;
        }

        .form button:hover {
            background-color: var(--blue);
        }
    </style>
</head>

<body>

    <form class="form">

        <h2>Registration Form</h2>

        <label for="name">Name</label>
        <input type="text" id="name" placeholder="Enter your name">

        <label for="email">Email</label>
        <input type="email" id="email" placeholder="Enter your email">

        <label for="password">Password</label>
        <input type="password" id="password" placeholder="Enter your password">

        <label for="date">Date of Birth</label>
        <input type="date" id="date">

        <label for="course">Choose Course</label>
        <select id="course">
            <option>HTML</option>
            <option>CSS</option>
            <option>JavaScript</option>
            <option>Python</option>
        </select>

        <label>Gender</label>
        <input type="radio" name="gender"> Male
        <input type="radio" name="gender"> Female

        <label>
            <input type="checkbox">
            I agree to the terms and conditions
        </label>

        <div class="form-section">
            This section demonstrates opacity.
        </div>

        <label for="message">Message</label>
        <textarea id="message" rows="4" placeholder="Write your message"></textarea>

        <button type="submit">Submit</button>

    </form>

</body>
</html>
"""

display(HTML(html))
```
